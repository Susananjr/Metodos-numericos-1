**Método de bisección para ecuaciones no lineales**

Para funciones no lineales (una función cuya gráfica no es una recta) de manera numérica se toma $f(x_r)=0$ dónde $x_r$ es una solución de la la ecuación, con dos tipos de soluciones posibles ya sean imaginarias o reales.
Existen varios métodos para resolver este tipo de función cerrados o abiertos en este caso, se aplicará un método cerrado siendo éste el método de mi bisección.
En este método partimos tomando una función $f(x)$ a la que se le quiere encontar una solución en el intervalo $[a,b]$, localizar una de las raíces del intervalo, al reducirlo de manera cíclica en dos partes cada vez más pequeñas hasta atrapar la raíz o estar más cercana a ella.


Primero importamos las librerias para ecuaciones matemáticas

```python
import numpy as np
import matplotlib.pyplot as plt

def f(x):
  return x**3 + 4*x**2 -10.0
```

Para la muestra del método usaremos la función $f(x)=x^3+4x^2-10$ en el intervalo $[0,2]$
La primera evaluación  de $f(x)$ es sobre los intervalos; uno debe ser positivo y el otro negativo para que nos garantice que la raíz estará en el eje $x$ (en caso contrario sería una solución imaginaría).

```python
0#para una visualización de la gráfica:
x=np.linspace(0,3)
plt.plot(x,f(x))
plt.title("Grafica $f(x)=x^3+4x^2-10$")
plt.axhline(y=0, color='red')
```

El proceso se basa en dividir en dos el intervalo, para la primera iteración (se realiza con los invervalos dados) $p_1=\frac{a_1+b_1}{2}$, con las condiciones:


*   Si $f(p_1)$ y $f(a_1)$ tienen el mismo signo, las coordenadas pasan a ser $a_2=p_1$ y $b_2=b_1$
*  Si $f(p_1)$ y $f(a_1)$ tienen signos opuestos, las coordenadas pasan a ser $a_2=a_1$ y $b_2=p_1$

El intercambio de los intervalos y las bisecciones dentro del programa se realizarán hasta un máximo de $n=100$ en caso de no encontrar una raíz con el error propuesto.
Si llega antes del número de iteraciones entonces encotraremos la raíz.



```python
#ingresa datos por entrada para los metodos la trabajar
a=1.0
b=2.0

tol= 0.0001 #tolerancia de decimales
nmax = 100 #n máximos para no crear un bucle infinito
niter = 0
error= 100.0
```

```python
#Evaluación de los intervalos
if f(a)*f(b)>0:
  print("El intervalo no es válido")
else:

     print( #estructura para el encabezado de tabla *recomendación* cuidar la identación, por experiencia puede no aparecer solo por este detalle
       f"{'iter':<6}{'a':<12}{'b':<12}{'p1 (Raíz)':<14}{'f(p1)':<14}{'error':<12}"
     )


     # guarda el valor de a y toma el valor inicial de manera temporal

     while error > tol and niter < nmax:  #condición para detener el algoritmo

       #error>tol si el error baja y se hace menor a tol, el ciclo ya alcanzó el límite y se detiene
       #al mismo tiempo si niter<nmax el número limitado de n para que no sea un bucle infinito
         niter += 1 # registro del contador
       #Calcular el punto medio
         p1 = (a+b)/2.0
         error = (b - a) / 2.0 #recalcula la parte sobrante del intervalo para tener la segunda comparación en la siguiente condición

         print( #Esta estructura expondrá el valor actualizado en cada ciclo de las variables en la izquierda con su límite de decimales
           f"{niter:<6}{a:<12.6f}{b:<12.6f}{p1:<14.6f}{f(p1):<14.5f}{error:<12.6f}"
         )

       #comprueba si el error actual es menor que 0.0001 para cortar la  iteración o continuarla con "or"
         if error<tol or f(p1)==0: #con el detalle de "==0" si encontramos la raíz exacta
                break
       #Evaluación de signos para actualizar el intervalo

         if f(p1)*f(b)<0:
           a = p1
         else:
           b = p1
       #Cuando se encuentra las raíz o terminan el número máximo de n iteraciones
```

```python
#COPIA Y REPETICIÓN DEL CÓDIGO PARA SU VISUALIZACIÓN JUNTO CON LA TABLA DE RESULTADOS


#Evaluación de los intervalos
if f(a)*f(b)>0:
  print("El intervalo no es válido")
else:

     while error > tol and niter < nmax:  #condición para detener el algoritmo

       #error>tol si el error baja y se hace menor a tol, el ciclo ya alcanzó el límite y se detiene
       #al mismo tiempo si niter<nmax el número limitado de n para que no sea un bucle infinito
         niter += 1 # registro del contador
       #Calcular el punto medio
         p1 = (a+b)/2.0
         error = (b - a) / 2.0 #recalcula la parte sobrante del intervalo para tener la segunda comparación en la siguiente condició

       #comprueba si el error actual es menor que 0.0001 para cortar la  iteración o continuarla con "or"
         if error<tol or f(p1)==0: #con el detalle de "==0" si encontramos la raíz exacta
                break
       #Evaluación de signos para actualizar el intervalo

         if f(p1)*f(b)<0:
           a = p1
         else:
           b = p1
       #Cuando se encuentra las raíz o terminan el número máximo de n iteraciones
```

Para el resultado, tendremos el número de iteraciones necesarias para llegar a la solución, la raíz aproximada y el valor de $f(x)$ en la raíz.

```python
print(f"Número de iteraciones: {niter}")
print(f"Raíz aproximada calculada: x={p1:.8f}") #.8f para redondear los decimales a ocho cifras
print(f"Valor de f(x) en la raíz: f({p1:.8f}) ={f(p1):.8f} ") #este número se puede aumentar o disminuir

```
