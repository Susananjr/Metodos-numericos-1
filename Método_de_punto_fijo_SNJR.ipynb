# MÉTODO DE PUNTO FIJO
El objetivo de este método es, dado un problema donde se busca hallar una raíz, para ecuaciones no lineales de la forma $f(x)=0$ en un intervalo $[a,b]$, cuando no es posible o conveniente despejar $x$ de forma algebraica directa.

De manera gráfica ya no se busca la raíz sobre el eje $x$, si no con la intersección entre una curva $y=g(x)$ y la recta de $45°(y=x)$, el punto donde ambas se intersectan se considera  **punto fijo** $p$ y al mismo timpo la **raíz**.

Condiciones oara la existencia de un punto fijo:


*   Si $g \in C[a,b]$ y $g(x) \in [a,b]$ para todas las $x \in [a,b]$, entonces $g$ tiene por lo menos un punto fijo en $[a,b]$.
*   Si, $g'(x)$ existe en $(a,b)$ y hay una constante positiva $k<1$ con $|g'(x)≤k|$, para todas las $x\in(a,b)$, entonces, existe exactamente un punto fijo en $[a,b]$.

Ya mencionado que hay ecuaciones de despejes complicados, lo que si podemos hacer son aproximaciones con cualquier grado específico de precisión.
a continuación el método (suponiendo que existe un punto fijo):

Con $n=1, 2, 3...$ para la sucesión de aproximaciones ${Pn}$ eligiremos un valor inicial $P_o$ cualquiera, dentro del intervalo.
Es necesario primero proponer ecuaciones de la forma $p=g(p)$.

Ej. considere la ecuación:
$x^3+4x^2-10=0$

Nuestra primera propocisión es:
$x=g_1(x)=x-x^3-4x^2+10$ y $P_o$=1.5 que es un punto dentro del intervalo.

Este punto $P_o$ lo evaluaremos en la ecuación $g_1=(x)$, el resultado de $g_1(P_0)=P_2$ y $P_2$ será el siguiente punto a evaluar.
Se realizará éste proceso consecutivamente mientras haya congruencia entre la serie y al mismo tiempo las evaluaciones tiendan a un número en común.



Funciones propuestas en código:

```python
import numpy as np
import matplotlib.pyplot as plt
import math

def f(x):
  return x**3 + 4*x**2 -10.0

def g1(x):
  return x-x**3 + 4*x**2 -10.0

def g2(x):
  return math.sqrt((10/x)-4*x)

def g3(x):
  return (1/2)*math.sqrt(10-x**3)

def g4(x):
  return math.sqrt(10/(4+x))

def g5(x):
  return x-((x**3+4*x**2-10)/(3*x**2+8*x))
```

```python
x=np.linspace(0,3)
plt.plot(x,f(x))
plt.plot(x,g1(x))
```

```python
#Código base, a partir de este código lo replicaremos adpatandolo a cada una de las funciones
a=1.0
b=2.0 #los intervalos propuestos
p0= 1.5 #como punto inicial dentro del intervalo
tol=0.0001
n0=100 #límite de iteraciones para no hacer un bucle infinito

#print(f"{'i':<6}{'pi':<14}{'g(pi)':<14}{'error':<16}") #formato para la impresión del encabezado de la tabla

i=1
p_antes=p0 #variable que actualizará el valor que genere para evaluarlo en el siguiente algoritmo
converge = False #CONDICION QUE NECESITÓ INVESTIGACIÓN ya que fue complicado que solo tuviera error con las iteraciones, se necesitó un
#límite en la cantidad de decimales para determinar que ya iba a converger

g_p0 = g1(p0) #evaluación del valor
#print(f"{0:<6}{p0:<14.8f}{g_p0:<14.8f}{'e':<16}")-- #formato para imprimir la actualización de los datos

while i<= n0: #bucle para las operaciones
      p= g1(p_antes)

      if p< a or p > b: #prueba si el valor generado está dentro del intervalo

        print("el valor se sale del intervalo")
        break


      error= abs(p-p_antes)
      gp= g1(p)

      #print(f"{i:<6}{p:<14.8f}{gp:<14.8f}{error:<16.8f}")--#impresion de los datos con los decimales propuestos Ej. .8 son: 0.00000001

      if error < tol: #evalua si error es mas pequeño que la tolerancia
        converge = True
        break #lo rompe en cuanto inclumpla la condicion

      p_antes=p #guarda el valor calculado para reutilizarlo en el siguiente ciclo
      i += 1 #iteracion para repetir el bucle

if converge:
  print(f"Punto fijo en {p:.8f} en {i} iteraciones")
else:
   print(f"Punto fijo falló, no converge dentro del intervalo")



```

```python
a=1.0
b=2.0
p0= 1.5
tol=0.0001
n0=100

print(f"{'i':<6}{'pi':<14}{'g(pi)':<14}{'error':<16}")

i=1
p_antes=p0
converge = False

g_p0 = g1(p0)
print(f"{0:<6}{p0:<14.8f}{g_p0:<14.8f}{'e':<16}")

while i<= n0:
      p= g1(p_antes)

      if p< a or p > b:

        print("el valor se sale del intervalo")
        break


      error= abs(p-p_antes)
      gp= g1(p)

      print(f"{i:<6}{p:<14.8f}{gp:<14.8f}{error:<16.8f}")

      if error < tol:
        converge = True
        break

      p_antes=p
      i += 1

if converge:
  print(f"Punto fijo en {p:.8f} en {i} iteraciones")
else:
   print(f"Punto fijo falló, no converge dentro del intervalo")
```
